```python
#make sure this is a file which contains gw (gaussian weights) - or you can calculate the gaussian weights yourself
print('define LatWeightAnyVar')
File_h0 = "/nobackup/pybl/cesm2/archive/b.e21.BWma1850.f19_g17.1yre0.001/atm/hist/b.e21.BWma1850.f19_g17.1yre0.001.cam.h0.0001-01.nc" #file name
DataSet_h0 = xr.open_dataset(File_h0,decode_times=False) #open the file and decode time as false

gw = DataSet_h0.gw.values

Pressure = DataSet_h0.lev.values
#%% define latitudinal weighting function

def LatWeightAnyVar(DS, time = False):
    try:
        DS = DS.mean(dim='lon')#take longitudinal mean
    except:
        print('')
    if (time == False):
        try:
            DS = DS.mean(dim='time')#take time mean
        except:
            print('')
    DS = gw*DS #multiply by gaussian weights
    try:
        DS = np.sum(DS,axis=1)/gw.sum() #weighted mean
    except:
        DS = np.sum(DS,axis=0)/gw.sum() #weighted mean 
    return DS #return a numpy array that has been modified above
```

```python
gw = DataSet_h0.gw.values


def LatWeightAnyVar_lat(DS,lat_low,lat_high,time = False): #Latitudinal weighting function for any latitudinal bands
    
    DS_lat=[]

    DS_gw=[]
    
    global count
    
    count = 0
    
    DS = DS.mean(dim='lon') #take longitudinal mean

    #DS = DS.mean(dim='time') #take time mean
        
    for i in range(len(DS.lat)):
        
        if DS.lat[i] >= lat_low and DS.lat[i] <= lat_high :
            
            DS_lat.append(DS[i])  # extract TS for certain lattitude
            
            count +=1 
            
    for i in range(len(DS_lat)): 
        
        DS_gw.append(gw[i]*DS_lat[i]) #multiply by gaussian weights
    
    DS_gw_final = np.sum(DS_gw)/gw[:count].sum()

    return DS_gw_final #return a numpy array that has been modified above
```

```python
from astropy.nddata import StdDevUncertainty, NDDataArray

x=list(range(3,39))
xrange=np.array(x)

y1_list=[y_e01_global_ice,y_e01_tropical_ice,y_e01_S_mid_ice,y_e01_N_mid_ice,y_e01_S_polar_ice,y_e01_N_polar_ice]

y2_list=[y_default_global_ice,y_default_tropical_ice,y_default_S_mid_ice, y_default_N_mid_ice,y_default_S_polar_ice,y_default_N_polar_ice]

# def mfit_param(x,y1_list,y2_list): # y1 is for e01, y2 is for default e
    
            
popt1=[]
popt2=[]
pcov1=[]
pcov2=[]

MLR_slope1=[]
MLR_slope2=[]

MLR_Serr1=[]
MLR_Serr2=[]

MLR_Perr1=[]
MLR_Perr2=[]
    
for i in range(len(y1_list)):
    
    popt1.append(cf(Multi_Reg, xrange, y1_list[i], p0=[1, 1, 1, 1, 1, 1])[0])
        
    popt2.append(cf(Multi_Reg, xrange, y2_list[i], p0=[1, 1, 1, 1, 1, 1])[0])
                
    pcov1.append(cf(Multi_Reg, xrange, y1_list[i], p0=[1, 1, 1, 1, 1, 1])[1])
        
    pcov2.append(cf(Multi_Reg, xrange, y2_list[i], p0=[1, 1, 1, 1, 1, 1])[1])
        
    MLR_slope1.append(popt1[i][1])
        
    MLR_slope2.append(popt2[i][1])
    
    MLR_Serr1.append((np.sqrt(np.diag(pcov1[i])))[1])
        
    MLR_Serr2.append((np.sqrt(np.diag(pcov2[i])))[1])
    
    MLR_Perr1.append(((np.sqrt(np.diag(pcov1[i])))[4])%(2*np.pi))
        
    MLR_Perr2.append(((np.sqrt(np.diag(pcov2[i])))[4])%(2*np.pi))
        
    
popt_sin1, pcov_sin1 = cf(sin,days/30,dist, p0=[1, 1, 1])
    
popt_sin2, pcov_sin2 = cf(sin,days_e01/30,dist_e01, p0=[1, 1, 1])
        
perr_sin1 = np.sqrt(np.diag(pcov_sin1))
    
perr_sin2 = np.sqrt(np.diag(pcov_sin2))
    
phase_e01=[]
phase_default=[]

# for i in range(len(popt1)):
#     print(popt1[i][2])
for i in range(len(popt1)):
    
    if popt1[i][2]>0:
        
        phase_e01.append(popt1[i][4]%(2*np.pi))
        
    if popt1[i][2]<0:
        
        phase_e01.append((popt1[i][4]+np.pi)%(2*np.pi))
        
    if popt2[i][2]>0:
        
        phase_default.append(popt2[i][4]%(2*np.pi))
            
    if popt2[i][2]<0:
        
        phase_default.append((popt2[i][4]+np.pi)%(2*np.pi))
        

if popt_sin1[1]>0:
        
    phase_esd_e01=(popt_sin1[2]%(2*np.pi))
          
if popt_sin1[1]<0:
        
    phase_esd_e01=(popt_sin1[2]+np.pi)%(2*np.pi)
        
if popt_sin2[1]>0:
        
    phase_esd_default=(popt_sin2[2]%(2*np.pi))
        
if popt_sin2[1]<0:
        
    phase_esd_default=((popt_sin2[2]+np.pi)%(2*np.pi))


phase_sf_default = phase_esd_default + np.pi
        
phase_sf_e01 = phase_esd_e01 + np.pi
    
lag_default=[]
lag_e01=[]
lag_default_err=[]
lag_e01_err=[] 

for i in range(len(phase_default)):
    #
    
    Phase_d_anderr      =   NDDataArray(phase_default[i], uncertainty=StdDevUncertainty(MLR_Perr2[i]))
    
    Phase_d_anderr_SF   =   NDDataArray(phase_sf_default, uncertainty=StdDevUncertainty(perr_sin2[2]))
    
    error_d = Phase_d_anderr.subtract(Phase_d_anderr_SF)
    
    #
    
    Phase_e_anderr      =   NDDataArray(phase_e01[i], uncertainty=StdDevUncertainty(MLR_Perr1[i]))
    
    Phase_e_anderr_SF   =   NDDataArray(phase_sf_e01, uncertainty=StdDevUncertainty(perr_sin1[2]))
    
    error_e = Phase_e_anderr.subtract(Phase_e_anderr_SF)
    #
    
    lag_default.append((phase_default[i]-phase_sf_default)%(2*np.pi))
    
    lag_default_err.append(error_d.uncertainty.array)
    #
    
    lag_e01.append((phase_e01[i]-phase_sf_e01)%(2*np.pi))
    
    lag_e01_err.append(error_e.uncertainty.array)
    

lag_default = [(x/(np.pi*2))*12 for x in lag_default]
lag_e01 = [(x/(np.pi*2))*12 for x in lag_e01]

print(type(phase_default))

print(phase_default)
print(lag_default_err)

print('MLR_slope1',MLR_slope1)
print('MLR_slope2',MLR_slope2)

print('MLR_Serr1',MLR_Serr1)
print('MLR_Serr2',MLR_Serr2)

print('lag_default',lag_default)
print('lag_e01',lag_e01)

print('lag_default_err',lag_default_err)
print('lag_e01_err',lag_e01_err)
          
# example data

x=[1,2,3,4,5,6]

y=lag_default
y1=lag_e01
# example variable error bar values
yerr = lag_default_err
yerr1=lag_e01_err



# First illustrate basic pyplot interface, using defaults where possible.
plt.figure(figsize=(8,6))
plt.errorbar(x, y, yerr, color='red',linestyle="None", marker='o', capsize=5, capthick=1, ecolor='black',label='$e=e_{\oplus}$')
plt.errorbar(x, y1, yerr1, color='green',linestyle="None", marker='o', capsize=5, capthick=1, ecolor='black',label='$e=0.1$')

#plt.title("Phase Lag of TS to Solar Forcing for $\epsilon=0$ Cases in Different Latitudinal Bands")
labels=['global','tropical','S_mid', 'N_mid','S_polar','N_polar']
plt.xticks(x, labels)
plt.xlabel("Latitudinal Bands",fontsize=15)
plt.ylabel("Phase Lag in Months",fontsize=15)
plt.legend()
plt.show()

#

x=[1,2,3,4,5,6]

plt.figure(figsize=(8,6))
plt.errorbar(x, MLR_slope2, MLR_Serr2, color='red',linestyle="None",marker='o', capsize=5, capthick=3, ecolor='black',label='$e=e_{\oplus}$')
plt.errorbar(x, MLR_slope1, MLR_Serr1, color='green',linestyle="None",marker='o', alpha=.3, fmt=':',capsize=5, capthick=3, ecolor='blue',label='$e=0.1$')

#plt.title("Fitted Trending Slopes of TS for $\epsilon=0$ Cases in Different Latitudinal Bands")
labels=['global','tropical','S_mid', 'N_mid','S_polar','N_polar']
plt.xticks(x, labels)
plt.xlabel("Latitudinal Bands",fontsize=15)
plt.ylabel("Trend Slope per Month",fontsize=15)
plt.legend()
plt.show()
```
