```python
# The function is used to latitudinally average variables, taking consideration of Earth's curvation factor
# as a guassian weight.

File_h0 = "/nobackup/pybl/cesm2/archive/b.e21.BWma1850.f19_g17.e0_obl0.001/atm/hist/b.e21.BWma1850.f19_g17.e0_obl0.001.cam.h0.0001-03.nc" #file name

DataSet_h0 = xr.open_dataset(File_h0,decode_times=False) #open the file and decode time as false

gw = DataSet_h0.gw.values

Pressure = DataSet_h0.lev.values

gw = DataSet_h0.gw.values


def LatWeightAnyVar_lat(DS,lat_low,lat_high,time = False): #Latitudinal weighting function for any latitudinal bands
    
    DS_lat=[]

    DS_gw=[]
    
    global count
    
    count = 0
    
    DS = DS.mean(dim='lon') #make zonal mean

    #DS = DS.mean(dim='time') #take time mean
        
    for i in range(len(DS.lat)): #make regional mean
        
        if DS.lat[i] >= lat_low and DS.lat[i] <= lat_high :
            
            DS_lat.append(DS[i])  # extract TS for certain lattitude
            
            count +=1 
            
    for i in range(len(DS_lat)): 
        
        DS_gw.append(gw[i]*DS_lat[i]) #multiply by gaussian weights
    
    DS_gw_final = np.sum(DS_gw)/gw[:count].sum()

    return DS_gw_final #return a numpy array that has been modified above
```

```python
# # Here I want to latitudinally-average land/ice/ocean fraction for N&S hemispheres
# # x-axis to be time, y-axis to be fraction, draw 3 types of lines on the same plot
# # need to consider guassian weighting because of spherical curvation of Earth

ICEFRAC_NH=[]
ICEFRAC_SH=[]
LNDFRAC_NH=[]
LNDFRAC_SH=[]
OCNFRAC_NH=[]
OCNFRAC_SH=[]

for i in range(len(ICEFRAC_obl0[4].ICEFRAC)):
    
    ICEFRAC_NH.append(LatWeightAnyVar_lat(ICEFRAC_obl0[4].ICEFRAC[i,:,:],lat_low=0,lat_high=90))
    
    ICEFRAC_SH.append(LatWeightAnyVar_lat(ICEFRAC_obl0[4].ICEFRAC[i],lat_low=-90,lat_high=0))
    
    LNDFRAC_NH.append(LatWeightAnyVar_lat(LANDFRAC_obl0[4].LANDFRAC[i],lat_low=0,lat_high=90))
    
    LNDFRAC_SH.append(LatWeightAnyVar_lat(LANDFRAC_obl0[4].LANDFRAC[i],lat_low=-90,lat_high=0))
    
    OCNFRAC_NH.append(LatWeightAnyVar_lat(OCNFRAC_obl0[4].OCNFRAC[i],lat_low=0,lat_high=90))
    
    OCNFRAC_SH.append(LatWeightAnyVar_lat(OCNFRAC_obl0[4].OCNFRAC[i],lat_low=-90,lat_high=0))
```

```python
plt.plot(OCNFRAC_SH,label='OCNFRAC_SH')
plt.plot(OCNFRAC_NH,label='OCNFRAC_NH')
plt.plot(LNDFRAC_SH,label='LNDFRAC_SH')
plt.plot(LNDFRAC_NH,label='LNDFRAC_NH')
plt.title("Ocean vs Land Fraction for e=0.4")

# plt.plot(ICEFRAC_NH,label='ICEFRAC_SH',linestyle='--')
# plt.plot(ICEFRAC_SH,label='ICEFRAC_SH',linestyle='--')
plt.legend()
plt.xlabel('time (month)')
plt.ylabel('fraction')
```
