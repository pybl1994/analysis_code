```python
n, bins, patches = plt.hist(df.eccentricity, bins=20, color='r', edgecolor='k', alpha=0.65)

plt.axvline(df.eccentricity.median(), color='k', linestyle='dashed', linewidth=1,label='median')

plt.legend()

plt.ylabel('Number of Planets')

plt.xlabel('Eccentricity')
```

```python
df.eccentricity.plot.hist(stacked=True, bins=20,color='r', edgecolor='k', alpha=0.65,label='relaxed')
plt.axvline(df.eccentricity.median(), color='k', linestyle='dashed', linewidth=1,label='relaxed,median')
df2.eccentricity.plot.hist(stacked=True, bins=20,color='g', edgecolor='k', alpha=0.65,label='orginal')
plt.axvline(df2.eccentricity.median(), color='k', linestyle='solid', linewidth=1,label='original,median')
plt.legend()
```

```python
df.eccentricity[~np.isnan(df.eccentricity)].plot.hist(stacked=True, bins=20,color='r', edgecolor='k', alpha=0.65,label='relaxed')
plt.axvline(df.eccentricity[~np.isnan(df.eccentricity)].median(), color='k', linestyle='dashed', linewidth=1,label='relaxed,median')
df2.eccentricity[~np.isnan(df2.eccentricity)].plot.hist(stacked=True, bins=20,color='g', edgecolor='k', alpha=0.65,label='orginal')
plt.axvline(df2.eccentricity[~np.isnan(df2.eccentricity)].median(), color='k', linestyle='solid', linewidth=1,label='original,median')
plt.legend()
```
