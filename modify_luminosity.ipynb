```python
import numpy as np
import xarray as xr


#%% WACCM Sun file
path = '/nobackup/phydrm/cesm2_inputdata//atm/cam/solar/'

solar_file = path+'SolarForcingCMIP6piControl_c160921.nc'#read in file

ds = xr.open_dataset(solar_file)#attahc file to dataset

ssi = ds['ssi'].isel(time=0) #define dataset from file

WACCM_Flux = ssi.values

WACCM_wavelength = ssi.wavelength.values

####################################

scale_factor_e04 = (1/1.1274092203200037)**2
scale_factor_e0  = (1/0.9999297316946573)**2

ds.ssi.values[0] = WACCM_Flux*scale_factor_e0; ds.ssi.values[1] = WACCM_Flux*scale_factor_e0

NewFile = '/nobackup/pybl/cesm2/Modified_SolarForcingCMIP6piControl_c160921_meanflux_circular_new.nc'

ds.to_netcdf(NewFile)

#%% Read the file in and check values

solar_file = NewFile #read in file

DS = xr.open_dataset(solar_file)#attahc file to dataset

ssi = DS['ssi'].isel(time=0) #define dataset from file

WACCM_Flux_TSI_e0 = ssi.values

from scipy.integrate import trapz

TSI_S0  = trapz(WACCM_Flux, WACCM_wavelength)/1000

TSI_e0 = trapz(WACCM_Flux_TSI_e0, WACCM_wavelength)/1000
print(TSI_e0)
print(TSI_e0/TSI_S0)
```
