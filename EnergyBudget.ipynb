```python
File_h0 = "/nobackup/pybl/cesm2/archive/b.e21.BWma1850.f19_g17.e0_obl0.001/atm/hist/b.e21.BWma1850.f19_g17.e0_obl0.001.cam.h0.0001-03.nc" #file name

DataSet_h0 = xr.open_dataset(File_h0,decode_times=False) #open the file and decode time as false

gw = DataSet_h0.gw.values

Pressure = DataSet_h0.lev.values

gw = DataSet_h0.gw.values

#%% define latitudinal weighting function

def LatWeightAnyVar(DS, time = False):
    try:
        DS = DS.mean(dim='lon')#take longitudinal mean
    except:
        print('')
    DS = gw*DS #multiply by gaussian weights
    try:
        DS = np.sum(DS,axis=1)/gw.sum() #weighted mean
    except:
        DS = np.sum(DS,axis=0)/gw.sum() #weighted mean 
    return DS #return a numpy array that has been modified above
```

```python
# Radiative timescale formula from Guendelman et al 2019

tau_IR = 0.1 # the infrared optical depth of the atmosphere,  
              # near-infrared regions, where the optical depth is of order 10−2
              # Observing Oceans in Tightly Packed Planetary Systems: Perspectives from Polarization
              # Modeling of the TRAPPIST-1 System Kopparla et al. (2018).
              # total aerosol optical depth is about 0.1 in US https://gml.noaa.gov/grad/surfrad/aod/
            
c_p    =  700 # heat capacity of air, ~ 700 J/KG/K

ps     = 100000 # surface pressure 1 Pa = 1 kg·m−1·s−2 1 hpa = 100 pa

g      = 9.8 # surface gravity 9.80665 m/s2.

sig_b  = 5.67 *10**(-8) # stefen-bolzmann constant  5.670374419...×10−8 W⋅m−2⋅K−4

T_e    = 255            # equilibrium temperature 255 K

tau = tau_IR * c_p * ps /(4*g*sig_b*T_e**3)

print('the estimated radiative timescale is (in days)', tau/(3600*24))

print('It is smaller than the orbital period of Earth, thus TS has similar structure to the insolation, seasonal response controlled \n by seasonal variability instead of the annual mean insolation')
```
